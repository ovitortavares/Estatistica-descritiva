## 4: Desvendando a Correlação Linear
Conceito:  

A correlação de Pearson é uma métrica matemática que mede a força e a direção da relação linear entre duas variáveis:

- Próximo de +1: Correlação positiva forte (quando uma sobe, a outra também sobe).

- Próximo de 0: Sem relação linear identificável.

- Próximo de -1: Correlação negativa forte (quando uma sobe, a outra desce).

Mapa de Calor (sns.heatmap): Converte essa tabela de números em uma matriz de cores para facilitar a leitura visual imediata.



### Exemplo Prático  

Cenário  
Uma empresa deseja verificar se existe relação entre a quantidade vendida e o faturamento.

```python
import pandas as pd

dados = {
    "Quantidade": [1, 2, 3, 4, 5],
    "Valor_Total": [100, 200, 300, 400, 500]
}

df = pd.DataFrame(dados)

correlacao = df["Quantidade"].corr(
    df["Valor_Total"]
)

print("Correlação:", correlacao)
```

Sempre que a quantidade aumenta, o valor total também aumenta na mesma proporção.

```python
import pandas as pd

dados = {
    "Quantidade": [1, 2, 3, 4, 5, 6],
    "Valor_Total": [120, 250, 280, 430, 510, 640]
}

df = pd.DataFrame(dados)

print(df.corr())
```

Como Interpretar o Resultado

|Resultado	|Interpretação|

|0,90 a 1,00	|Correlação muito forte|
|0,70 a 0,89	|Correlação forte|
|0,40 a 0,69	|Correlação moderada|
|0,10 a 0,39	|Correlação fraca|
|0,00 a 0,09	|Quase inexistente|

### Enunciado e Pergunta da Análise  

Calcule a correlação entre todas as categorias de produtos e gere um mapa de calor com Seaborn.  
Qual produto tem a correlação mais forte com o faturamento total?

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from pathlib import Path

# Caminho dinâmico para a pasta 'dados' e o arquivo 'tabela.csv'
caminho = Path("..") / "DATA" / "dados_vendas.csv"

# 1. Carregar o arquivo CSV
df = pd.read_csv(caminho)

# 1. Calcular a matriz de correlação das colunas numéricas
corr = df[['Smartphones', 'Notebooks', 'Acessorios', 'Total_Vendas_Mil']].corr() # Calcula a correlação entre as colunas numéricas do DataFrame 
# .corr() retorna a matriz de correlação, que mostra como as variáveis estão relacionadas entre si. 

# 2. Exibir o mapa de calor
sns.heatmap(corr, annot=True, cmap='Blues')

plt.title('Matriz de Correlação')
plt.show()
```

Leitura Analítica:

- Smartphones e Faturamento: 0,98 (relação quase perfeita).
- Acessórios e Faturamento: 0,97 (relação extremamente alta).
- Notebooks e Faturamento: 0,94 (relação alta, mas ligeiramente menor).

## A Regra de Ouro da EDA:  
Correlação Não Implica CausalidadeEste é o conceito conceitual mais cobrado em Business Intelligence para iniciantes:

- Duas métricas podem crescer juntas (ter correlação alta, como Smartphones e Faturamento), mas isso não prova automaticamente que uma é a única causa da outra.

- Pode existir uma terceira variável oculta (ex.: investimento geral em marketing ou o crescimento do poder de compra no semestre) impulsionando ambas simultaneamente.

## Bônus:

Construção do Dashboard Executivo (Painel 2x2)  
Enunciado / Pergunta de BI:  
Consolide em uma única tela gerencial a evolução do faturamento, o comparativo de volume entre as categorias, a reta de tendência de vendas e a matriz de correlação do primeiro semestre.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from pathlib import Path

# Caminho dinâmico para a pasta 'dados' e o arquivo 'tabela.csv'
caminho = Path("..") / "DATA" / "dados_vendas.csv"

# 1. Carregar o arquivo CSV
df = pd.read_csv(caminho)

sns.set_theme(style='whitegrid', palette='Blues_r')

# Criar a grade do dashboard (2 linhas x 2 colunas)
fig, axes = plt.subplots(2, 2, figsize=(14, 9))
fig.suptitle(
    'Dashboard Executivo de Performance Comercial - 1º Semestre',
    fontsize=16,
    fontweight='bold',
    y=0.98,
)

# 1. Gráfico de Linha: Faturamento Mensal (Evolução Temporal)
sns.lineplot(
    data=df,
    x='Mes',
    y='Total_Vendas_Mil',
    marker='o',
    color='#08519c',
    linewidth=2.5,
    ax=axes[0, 0],
)
axes[0, 0].set_title('1. Faturamento Mensal (em R$ Mil)', fontweight='bold')
axes[0, 0].set_xlabel('Mês')
axes[0, 0].set_ylabel('R$ Mil')

# 2. Gráfico de Barras Agrupadas: Comparativo de Categorias
df_long = df.melt(
    id_vars=['Mes'],
    value_vars=['Smartphones', 'Notebooks', 'Acessorios'],
    var_name='Categoria',
    value_name='Unidades',
)
sns.barplot(
    data=df_long,
    x='Mes',
    y='Unidades',
    hue='Categoria',
    palette='Blues_d',
    ax=axes[0, 1],
)
axes[0, 1].set_title('2. Volume de Unidades por Categoria', fontweight='bold')
axes[0, 1].set_xlabel('Mês')
axes[0, 1].set_ylabel('Unidades')
axes[0, 1].legend(title='Linha de Produto')

# 3. Gráfico de Regressão: Dispersão + Linha de Tendência Linear
sns.regplot(
    data=df,
    x='Smartphones',
    y='Total_Vendas_Mil',
    color='#2171b5',
    scatter_kws={'s': 60},
    ax=axes[1, 0],
)
axes[1, 0].set_title(
    '3. Tendência: Smartphones vs Faturamento Total', fontweight='bold'
)
axes[1, 0].set_xlabel('Unidades de Smartphones')
axes[1, 0].set_ylabel('Faturamento (R$ Mil)')

# 4. Mapa de Calor: Matriz de Correlação
corr = df[
    ['Smartphones', 'Notebooks', 'Acessorios', 'Total_Vendas_Mil']
].corr()
sns.heatmap(
    corr,
    annot=True,
    cmap='Blues',
    fmt='.2f',
    cbar=False,
    square=True,
    ax=axes[1, 1],
)
axes[1, 1].set_title('4. Matriz de Correlação Multivariada', fontweight='bold')

plt.tight_layout()
plt.show()
```
